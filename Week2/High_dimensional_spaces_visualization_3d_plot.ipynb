```
%matplotlib inline
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
from matplotlib import animation
import numpy as np

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.set_xlim(-1,1)
ax.set_ylim(-1,1)
ax.set_zlim(-1,1)
ax.view_init(elev=20., azim=70)
ax.set_facecolor('xkcd:white')

x, y, z = np.zeros((3,3))
u, v, w = np.array([[1,],[1,],[1, ]])

ax.quiver(x,y,z,u,v,w,arrow_length_ratio=0.1)
plt.gca().xaxis.set_major_locator(plt.MultipleLocator(0.5))
plt.gca().yaxis.set_major_locator(plt.MultipleLocator(0.5))
plt.gca().zaxis.set_major_locator(plt.MultipleLocator(0.5))

# plt.show()

def rotate(angle):
    ax.view_init(azim=angle)

rot_animation = animation.FuncAnimation(fig, rotate, frames=np.arange(0, 90, 2),interval=100)
rot_animation.save('rotation.gif', dpi=80, writer='pillow')
```
